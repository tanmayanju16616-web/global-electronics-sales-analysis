### Convert Date Columns

**Q1.5** - Convert all date-related columns to a standardised datetime format. **(2 marks)**
* **Hint** - Use datetime function 

```python
customers["Birthday"] = pd.to_datetime(customers["Birthday"])

exchange_rates['Date'] = pd.to_datetime(exchange_rates['Date'])

sales['Order Date'] = pd.to_datetime(sales['Order Date'])
sales['Delivery Date'] = pd.to_datetime(sales['Delivery Date'],errors='coerce')

stores["Open Date"] = pd.to_datetime(stores["Open Date"])
```

### Transform Numerical Columns

**Q1.6** - Transform object-type columns containing numeric information into numerical data types for analysis. **(2 marks)**
* **Hint** - Use Replace function 

```python
products['Unit Cost USD'] = pd.to_numeric(products['Unit Cost USD'].replace({'[$,]': '', '^$': None}, regex=True), errors='coerce')

products['Unit Price USD'] = pd.to_numeric(products['Unit Price USD'].replace({'[$,]': '', '^$': None}, regex=True), errors='coerce')
```

### Join the Tables 

```python
# Merge FX rates
sales_fx = sales.merge(
    exchange_rates,
    left_on=['Order Date', 'Currency Code'],
    right_on=['Date', 'Currency'],
    how='left'
)

# Merge master data
sales_full = (
    sales_fx
    .merge(customers, on='CustomerKey', how='left')
    .merge(products, on='ProductKey', how='left')
    .merge(stores, on='StoreKey', how='left')
)
```

* A consolidated, analysis-ready dataset (sales_full) containing transactional, customer, product, store, and currency information in a single table.

### Normalise the Revenue

```python
# Create USD-normalised revenue
sales_full['Revenue_USD'] = (
    sales_full['Quantity'] *
    sales_full['Unit Price USD'] *
    sales_full['Exchange']
)

```

* Revenue was converted to USD to enable consistent cross-country financial analysis.

*  **Ans. 3.1.1** - Revenue per square meter is a suitable metric for assessing a store's revenue efficiency relative to its physical size.

```python
physical_stores = sales_full[sales_full['Square Meters'] > 0]

store_efficiency = physical_stores.groupby('StoreKey').agg({
    'Revenue_USD': 'sum',
    'Square Meters': 'first'
})

store_efficiency['Revenue_per_sqm'] = (
    store_efficiency['Revenue_USD'] / store_efficiency['Square Meters']
)
```

 * **Insights** - Some stores generate high revenue per square meter and some large stores have low efficiency. So, store size does not always lead to higher revenue.

 * **Recommendations** -
1. Consider downsizing large inefficient stores to improve performance.
2. Study high- performing stores and apply their strategies to others.

**Note** - Give atleast 1 insight and 1 recommendation


* **Metric** - Month-on-Month(MoM) Revenue growth(%)
* **MoM Growth(%)** = __((Current month revenue - Previous month revenue)/ Previous month revenue *100)__
 - We are going to use this MoM revenue growth(%) metric to measure the percent change in revenue between consecutive months for each store.

```python
sales_full["Month"] =sales_full['Order Date'].dt.to_period('M')
monthly_revenue_store= sales_full.groupby(['StoreKey','Month'])['Revenue_USD'].sum()
mom_growth_store = monthly_revenue_store.groupby(level=0).pct_change()*100
mom_growth_store = mom_growth_store.reset_index(name = 'MoM Revenue Growth(%)')
```

* **Insights** - 1. We can see variability in our month-on-month revenue growth, which shows unstable and inconsistent performance of the stores across countries.
2. We can see the seasonal pattern in the data. Revenue seems to increase during May–July, while March–April months are common low-performing months across multiple countries.
3. Store performance differs across countries. European markets show higher growth but with more fluctuations, whereas Canada has lower but more stable growth, indicating differences in customer demand across regions.

* **Recommendations** - 1. Improve sales strategies during low-performing months to reduce the drop in revenue.
2. Leverage high-performing months (May–July) by planning promotions and marketing campaigns.
3. Use more reliable metrics like median or filtered data to avoid the impact of extreme values on analysis.

### Perform Store Age–Based Segmentation

* **Q3.3.1 and 3.3.2** - Calculate the age of the stores in 5 buckets ('0–5 years', '5–10 years', '10–15 years', '15–20 years', '20+ years'). **(3 marks)**
* **Q3.3.3** - Calculate the total revenue in each bucket (write a Python code). **(2 marks)**
* **Q3.3.4** - Calculate the average revenue per store in each bucket (write a Python code). **(2 marks)**
* **Q3.3.5** - What insights and recommendations can be derived from this metric? **(2 marks)**

**Note** - Give atleast 1 insight and 1 recommendation


```python
sales_full['Store Age'] = (pd.Timestamp.today() - sales_full['Open Date']).dt.days / 365

bins = [0, 5, 10, 15, 20, float('inf')]
labels = ['0–5 years', '5–10 years', '10–15 years', '15–20 years', '20+ years']

sales_full['Age Bucket'] = pd.cut(sales_full['Store Age'], bins=bins, labels=labels)
```

```python
sales_full.groupby('Age Bucket',observed=False)['Revenue_USD'].sum().reset_index()
```

```python
rev_store = sales_full.groupby(['StoreKey', 'Age Bucket'], observed=True)['Revenue_USD'].sum()

avg_rev_bucket = rev_store.groupby('Age Bucket',observed=False).mean().sort_index().reset_index()
```

* **Insights**- From the analysis, we can conclude that most of the revenue is generated from the stores that are 10–20 years old, so, these stores are performaing better. Stores older than 20 years have the highest average revenue, but they are only few, their overall impact is smaller. Stores in the 5–10 year range are performing relatively lower. Overall, revenue seems to increase as stores grow older.

* **Recommendations** - Focus on strengthening 10–20 year old stores since most of the revenue is generated from these stores. Study the performance of 20+ year stores and apply those strategies elsewhere. Additionally, more support is needed to 5–10 year stores to improve their performance.
